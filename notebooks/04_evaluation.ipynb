# Notebook 04: TounsiEval Benchmark Evaluation

Visualizes benchmark results across the 14 linguistic categories, comparing generated outputs and human evaluation rubrics.

```python
import json
from pathlib import Path
import pandas as pd

eval_files = sorted(Path('../evaluation_results').glob('eval_report_*.json'))
if eval_files:
    latest = eval_files[-1]
    with open(latest, 'r', encoding='utf-8') as f:
        data = json.load(f)
    df = pd.DataFrame(data['results'])
    print(f'Loaded report from {latest.name}: {len(df)} samples')
    display(df[['category', 'input', 'model_output']].head(10))
else:
    print('No evaluation reports found. Run `python scripts/evaluate.py`.')
```
