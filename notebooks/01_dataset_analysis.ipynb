# Notebook 01: Tunisian Dataset Analysis

Analyzes document distributions, script ratios (Arabic vs Arabizi vs Code-Switching), and vocabulary statistics for TounsiLM.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt

# Load dataset report
report_path = Path('../data/manifests/dataset_report.json')
if report_path.exists():
    with open(report_path, 'r', encoding='utf-8') as f:
        report = json.load(f)
    print('Dataset Summary:')
    for k, v in report.items():
        print(f'  {k}: {v}')
else:
    print('Run `python scripts/deduplicate.py` to generate report.')
```
