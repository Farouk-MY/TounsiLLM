# Notebook 02: Tokenizer Inspection & Compression Audit

Evaluates subword pieces, compression ratios (chars/token), and roundtrip fidelity for Arabic Derja, Arabizi, and French code-switching.

```python
import sys
sys.path.insert(0, '../src')
from tounsilm.tokenizer.tokenizer import TounsiTokenizer

tok = TounsiTokenizer('../data/tokenizer/tounsilm_tokenizer.model')
print(f'Vocab size: {tok.vocab_size:,}')

samples = [
    'شنوة تعمل توّا؟',
    '3lech ma jitich lbare7?',
    'لازم نعمل update للـapplication.',
]

for s in samples:
    info = tok.inspect(s)
    print(f"{s} -> {info['num_tokens']} tokens ({info['chars_per_token']:.2f} chars/tok) -> {info['pieces']}")
```
